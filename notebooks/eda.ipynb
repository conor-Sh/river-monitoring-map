```python
import requests
import pandas as pd

BASE = "https://environment.data.gov.uk/flood-monitoring"
```

```python
# 1. Get stations
r = requests.get(f"{BASE}/id/stations", params={"parameter": "level", "_limit": 50})
r.raise_for_status()
stations = pd.DataFrame(r.json()["items"])
print(stations.columns)          # check the real variable names 
print(stations[["label", "riverName", "lat", "long", "stationReference"]].head())

# 2. Recent readings 
ref = stations.loc[0, "stationReference"]
r = requests.get(f"{BASE}/id/stations/{ref}/readings",
                 params={"_sorted": "", "_limit": 200})
readings = pd.DataFrame(r.json()["items"])
print(readings[["dateTime", "value"]].head())
```
